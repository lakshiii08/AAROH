## 1. Generate 4-Step Personalized Remedial Learning Pathway

```python
pathway = pipeline.generate_remedial_pathway(
    student_id='student_sukru_demo',
    student_name='Sukru Mandavi',
    concept_code='EVS-G3-WAT-01',
    weak_bloom_level='Understand',
    target_language='hi',
)

print(f'Pathway ID: {pathway.pathway_id}')
print(f'Student: {pathway.student_name} | Concept: {pathway.concept_name}')
print(f'Status: {pathway.status}\n')
for s in pathway.steps:
    print(f'Step {s.step_number} [{s.step_type.upper()}]: {s.title}')
    print(f'   Instruction: {s.instruction}')
    if s.resource_url:
        print(f'   Resource URL: {s.resource_url}\n')
```

## 2. Generate and Dispatch Mother-Tongue Voice Advisory for Parents
