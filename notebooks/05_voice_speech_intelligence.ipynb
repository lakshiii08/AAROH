# AAROH - Phase 5: Voice & Speech Intelligence Layer

This notebook demonstrates the complete Phase 5 pipeline:
1. **Text-to-Speech (TTS) Synthesis**: Converts simplified stories and lesson scripts into audio tracks in mother tongue.
2. **Multi-Track Voice Lesson Package**: Generates explanation, story, and voice recap tracks for a concept.
3. **Speech-to-Text (STT) for Oral Quizzes**: Transcribes spoken answers from primary school students.
4. **Oral Assessment & Spoken Feedback**: Grades oral responses with fuzzy matching and generates verbal encouragement tracks.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from src.aaroh.pipeline.voice_intelligence_pipeline import VoiceIntelligencePipeline

pipeline = VoiceIntelligencePipeline()
print('Phase 5 Voice Intelligence Pipeline initialized successfully!')
```

## 1. Synthesize Custom Mother-Tongue Audio Explanation

```python
text = 'Namaste doston! Mitti ka matka paani ko thanda kaise rakhta hai? Aao milke dekhte hain.'
segment = pipeline.synthesize_custom_text(text=text, section_name='intro', target_language='hi')

print(f'Generated Audio: {segment.audio_file_path}')
print(f'Duration: {segment.duration_seconds}s | Voice: {segment.voice_id}')
```

## 2. Generate Full Multi-Track Voice Lesson for a Concept

```python
voice_pkg = pipeline.generate_voice_lesson(
    concept_code='EVS-G3-WAT-01',
    target_language='hi',
    target_dialect='gon',
    grade_level=3,
)

print(f'Concept: {voice_pkg.concept_name} ({voice_pkg.concept_code})')
print(f'Total Audio Duration: {voice_pkg.total_duration_seconds}s\n')
for s in voice_pkg.segments:
    print(f'Track [{s.section_name}]: {Path(s.audio_file_path).name} ({s.duration_seconds}s)')
```

## 3. Student Oral Quiz Assessment & Spoken Audio Feedback

```python
# Simulate student oral response
dummy_audio = pipeline.config_manager.get_voice_config().audio_output_dir / 'student_ans.wav'
pipeline.tts._generate_local_wav(dummy_audio, 'matke ka paani bhaap banke thanda hota hai', 3.0)

oral_result = pipeline.process_oral_quiz_answer(
    quiz_id='quiz_demo_1',
    question_id='Q2',
    student_id='student_soma_01',
    audio_file_path=str(dummy_audio),
    expected_answer='matka',
    target_language='hi',
)

print(f'Transcribed Speech: {oral_result.transcribed_speech}')
print(f'Graded Result: {oral_result.is_correct} (Confidence: {oral_result.confidence_score})')
print(f'Verbal Feedback Script: {oral_result.feedback_audio_script}')
print(f'Feedback Audio Track: {oral_result.feedback_audio_path}')
```
